```julia
using MLJ
using DataFrames
using CSV
using Plots
```

```julia
# Generate half moons with a given total number of points
coords, ids = make_moons(1000, noise=0.05)

# Format data in DataFrame object
moonDataset = DataFrame(coords)
moonDataset.moon_id = ids
rename!(moonDataset, :x1 => :x, :x2 => :y)  # Rename "x1" and "x2" to "x" and "y" for clarity

# Save as .csv
parentFolder = dirname(@__DIR__)    # Save in "Datasets" folder
CSV.write(parentFolder * "/HalfMoonDataset.csv", moonDataset)
```

```julia
# Plot data

# Separate points by which half moon they belong to
moon0Dataframe = subset(moonDataset, :moon_id => ByRow(x -> x == 0))
moon1Dataframe = subset(moonDataset, :moon_id => ByRow(x -> x == 1))

p = scatter(moon0Dataframe.x, moon0Dataframe.y; label = "ID = 0")
scatter!(moon1Dataframe.x, moon1Dataframe.y; label = "ID = 1")
xaxis!("x")
yaxis!("y")
title!("Half Moon Dataset")
display(p)
```
